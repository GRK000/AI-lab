# Training A Network From Scratch

This notebook is a lightweight guided entry point for the NumPy neural network core.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'from-scratch'))

import numpy as np
from neural_core import DenseLayer, NeuralNetwork
```

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=np.float32)
y = np.array([0, 1, 1, 0])

model = NeuralNetwork(
    layers=[DenseLayer(8, 'tanh'), DenseLayer(1, 'sigmoid')],
    problem_type='binary',
    learning_rate=0.2,
    max_epochs=2500,
    batch_size=4,
    random_state=21,
)
model.fit(X, y)
model.score(X, y), model.predict(X)
```
